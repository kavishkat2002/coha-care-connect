# Automated Eye Cancer & Retinal Disease Detection
Deep Convolutional Neural Networks (ConvNets) based eye cancer and disease detection, trained on 1500 true colour 24-bit RGB fundus images across 26 categories.

### Supported Datasets:
1. **STARE** – STructured Analysis of the Retina
2. **DRIVE** – Digital Retinal Images for Vessel Extraction
3. **Messidor (base11)**
4. **Kaggle** – 1000 Fundus Images & Glaucoma Detection
5. **DRHAGIS** – Diabetic Retinopathy, Hypertension, AMD, and Glaucoma ImageS
6. **HRF** – High-Resolution Fundus


```python
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np

# 1. Define Deep ConvNet Architecture for Fundus Analysis
def build_fundus_convnet(input_shape=(256, 256, 3), num_classes=26):
    model = models.Sequential()
    
    # Convolutional Layers
    model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=input_shape))
    model.add(layers.MaxPooling2D((2, 2)))
    
    model.add(layers.Conv2D(64, (3, 3), activation='relu'))
    model.add(layers.MaxPooling2D((2, 2)))
    
    model.add(layers.Conv2D(128, (3, 3), activation='relu'))
    model.add(layers.MaxPooling2D((2, 2)))
    
    model.add(layers.Conv2D(256, (3, 3), activation='relu'))
    model.add(layers.MaxPooling2D((2, 2)))
    
    # Dense Classification Head
    model.add(layers.Flatten())
    model.add(layers.Dense(512, activation='relu'))
    model.add(layers.Dropout(0.5))
    model.add(layers.Dense(num_classes, activation='softmax')) # 26 Categories
    
    return model

model = build_fundus_convnet()
model.summary()
```

```python
# 2. Optimizer Selection & Model Compilation
# We test MBGD, Adagrad, RMSProp, but best loss minimization is achieved via Adam with learning_rate=1e-5
adam_optimizer = optimizers.Adam(learning_rate=1e-5)

model.compile(optimizer=adam_optimizer,
              loss='categorical_crossentropy',
              metrics=['accuracy', tf.keras.metrics.AUC()])
```

```python
# 3. Data Augmentation & Batch Loading
# Augmented fundus images in batches of size 32
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
    zoom_range=0.15,
    fill_mode='nearest'
)

# Simulated dataset initialization (1500 true colour images)
BATCH_SIZE = 32
TOTAL_IMAGES = 5304 # Number after augmentation
STEPS_PER_EPOCH = TOTAL_IMAGES // BATCH_SIZE # 165 iterations

print(f"Training initiated. Batch Size: {BATCH_SIZE}. Iterations per epoch: {STEPS_PER_EPOCH}")
```
